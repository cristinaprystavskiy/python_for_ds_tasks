Бачимо, що учасники програми лояльності (Member) витрачають більше, тому бізнесу варто розвивати цю програму, пропонувати бонуси, акції та заохочувати звичайних клієнтів приєднуватися до неї.

####5. **Якому способу оплати віддають перевагу різні типи клієнтів?**

```python
payment_by_customer = df.pivot_table(
    index='Customer type',
    columns='Payment',
    values='Invoice ID',
    aggfunc='count'
)

payment_by_customer
```

```python
print(f"Найпопулярніший спосіб оплати: {payment_by_customer.idxmax(axis=1)}")
```

Виявлено, що учасники програми лояльності (Member) найчастіше оплачують покупки кредитною карткою (Credit card), тоді як звичайні клієнти (Normal) віддають перевагу електронному гаманцю (Ewallet).

Отже, магазин може врахувати ці особливості під час розробки маркетингових пропозицій. Наприклад, для учасників програми лояльності пропонувати бонуси за безготівкову оплату карткою, а для звичайних клієнтів — заохочення за використання електронного гаманця. Це допоможе підвищити зручність оплати та лояльність покупців.

####6. **У які дні тижня супермаркет отримує найбільший дохід?**

```python
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
df['Day'] = df['Date'].dt.day_name()

days_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

sales_by_day = df.groupby('Day').agg(
    invoices=('Invoice ID', 'count'),
    revenue=('Total', 'sum')
).reindex(days_order)

print(sales_by_day)

print(f"Найбільший дохід отримано в: {sales_by_day['revenue'].idxmax()}")

print(f"Найменший дохід отримано в: {sales_by_day['revenue'].idxmin()}")
```

Найбільший дохід супермаркет отримує в суботу (56 120,81), а найменший — у понеділок (37 899,08). Також у суботу зафіксовано найбільшу кількість покупок — 164, тоді як у понеділок — 125.

На основі отриманих результатів супермаркет може приділити більше уваги підготовці до вихідних: забезпечити достатню кількість товарів та персоналу. У понеділок доцільно запроваджувати спеціальні акції та знижки, щоб стимулювати попит і збільшувати кількість покупок.

```python
sales_by_day['average_check'] = (sales_by_day['revenue'] / sales_by_day['invoices'])

print(sales_by_day)
```
